## Статистический анализ результатов исследования

```python
# функция для подсчета количества ответов и их процента
def answers_counts(column_name):
    results = (answers[column_name]
               .value_counts()
               .reset_index()
               .rename(columns={column_name: 'Ответ'})
               .rename(columns={'count': 'Количество'})  
               .assign(Процент=lambda x: (x['Количество'] / x['Количество'].sum() * 100)))
    return(results)
```

### 1.2 В какой мере вас интересует то, что происходит в Вашем поселении?

```python
# уникальные ответы
answers['1.2 В какой мере вас интересует то, что происходит в Вашем поселении?'].unique()
```

Создадим категории ответов с числовыми значениями, чтобы сравнить уровень интереса жителей по различных категориям (Сельское поселение, Пол, Возраст, Образование, Сфера деятельности, Тип семьи, Фин. состояние).

```python
# функция для преобразования текстовых ответов в числовые значения
def interest_index(response):
    if response == 'очень волнует':
        return 1
    elif response == 'постольку поскольку':
        return 0
    elif response == 'не интересует совсем':
        return -1
    else:
        return None  # на случай, если ответ не соответствует ожидаемым
```

```python
answers['1.2 индекс В какой мере вас интересует то, что происходит в Вашем поселении?'] = \
answers['1.2 В какой мере вас интересует то, что происходит в Вашем поселении?'].apply(interest_index)
```

```python
answers_counts('1.2 В какой мере вас интересует то, что происходит в Вашем поселении?')
```

### 1.3 Как Вы в целом оцениваете качество жизни в Вашем поселении?

```python
# уникальные ответы
answers['1.3 Как Вы в целом оцениваете качество жизни в Вашем поселении?'].unique()
```

```python
# функция для преобразования текстовых ответов в числовые значения
def quality_life(response):
    if response == 'низкое':
        return -1
    elif response == 'ниже среднего':
        return -0.5
    elif response == 'среднее':
        return 0
    elif response == 'выше среднего':
        return 0.5
    elif response == 'высокое':
        return 1
    else:
        return None  # на случай, если ответ не соответствует ожидаемым
```

### 1.4 Как Вы считаете, учитывается ли Ваше мнение при принятии решений о развитие Вашего поселения?

```python
# уникальные ответы
answers['1.4 Как Вы считаете, учитывается ли Ваше мнение при принятии решений о развитие Вашего поселения?'].unique()
```

```python
# функция для преобразования текстовых ответов в числовые значения
def importance_opinion(response):
    if response == 'да':
        return 1
    elif response == 'затрудняюсь ответить':
        return 0
    elif response == 'нет':
        return -1
    else:
        return None  # на случай, если ответ не соответствует ожидаемым
```

```python
answers['1.4 индекс Как Вы считаете, учитывается ли Ваше мнение при принятии решений о развитие Вашего поселения?'] = \
answers['1.4 Как Вы считаете, учитывается ли Ваше мнение при принятии решений о развитие Вашего поселения?'].apply(importance_opinion)
```

```python
column_name = '1.4 индекс Как Вы считаете, учитывается ли Ваше мнение при принятии решений о развитие Вашего поселения?'

results = (answers[column_name]
           .value_counts()
           .reset_index()
           .rename(columns={column_name: 'Ответ'})
           .rename(columns={'count': 'Количество'})  
           .assign(Процент=lambda x: (x['Количество'] / x['Количество'].sum() * 100)))
print(results)
```

#### Выводы
- 43.7% жителей воздержались от ответа, это довольно высокая доля респондентов, которые либо не имеют мнения по этому вопросу, либо не хотят или не могут выразить его. Это может указывать на: низкий уровень информированности о процессах принятия решений в местном управлении, непонимание значения вопроса или его актуальности для них, cкептическое или равнодушное отношение к тому, как их мнение может повлиять на развитие поселения;
- 37.3% ответили, что их мнение не учитывается, эта группа выражает недовольство и чувство исключенности из процессов принятия решений. Такой вывод говорит о: cуществующих проблемах в коммуникации между жителями и местными властями, ощущении безразличия со стороны власти к мнению жителей, возможной необходимости в улучшении механизма обратной связи или участия населения в процессе принятия решений.
- 19% считают, что их мнение учитывается. Этот процент относительно мал и может свидетельствовать о том, что лишь небольшая часть общественности уверена в том, что их мнения влияют на развитие поселения.

На основании этих данных можно сделать вывод о наличии значительного недовольства среди жителей по поводу отсутствия учёта их мнения.
